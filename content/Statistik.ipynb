```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

# Zufallsdaten generieren (echte "Experimente")

# Binomial: 10 Münzwürfe mit p=0.5 (1000 Wiederholungen)
binom_data = np.random.binomial(n=10, p=0.5, size=1000)

# Poisson: Anzahl Kunden pro Stunde bei λ=4 (1000 Stunden)
poisson_data = np.random.poisson(lam=4, size=1000)

# Normal: Körpergrößen Frauen (normalverteilt, z. B. μ=165, σ=6 cm)
normal_data = np.random.normal(loc=165, scale=6, size=10000)

# Exponential: Zeit bis nächster Kunde (λ=15)
expon_data = np.random.exponential(scale=15, size=10000)

# Plot
plt.figure(figsize=(12, 8))

# Binomialverteilung (Histogramm)
plt.subplot(2, 2, 1)
plt.hist(binom_data, bins=range(0, 11), align='left', rwidth=0.8)
plt.title("Binomialverteilung\nExperiment: 10-mal Münzwurf, P(Kopf)=0.5")
plt.xlabel("Anzahl Erfolge (Kopf)")
plt.ylabel("Häufigkeit")

# Poissonverteilung (Histogramm)
plt.subplot(2, 2, 2)
plt.hist(poisson_data, bins=range(0, 16), align='left', rwidth=0.8)
plt.title("Poissonverteilung\nExperiment: Kunden pro Stunde bei λ=4")
plt.xlabel("Anzahl Ereignisse")
plt.ylabel("Häufigkeit")

# Normalverteilung (Histogramm)
plt.subplot(2, 2, 3)
plt.hist(normal_data, bins=30, density=True)
plt.title("Normalverteilung\nExperiment: Körpergröße (μ=165cm, σ=6cm)")
plt.xlabel("x")
plt.ylabel("Dichte")

# Exponentialverteilung (Histogramm)
plt.subplot(2, 2, 4)
plt.hist(expon_data, bins=30, density=True)
plt.title("Exponentialverteilung\nExperiment: Zeit bis nächster Kunde (λ=15m)")
plt.xlabel("Zeit (Minuten)")
plt.ylabel("Dichte")

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import norm

# CSV-Datei einlesen (nach Upload in JupyterLite)
df_komplett = pd.read_csv("groesse_gewicht.csv")
df = df_komplett[df_komplett["Geschlecht"] == 0]

# Zwei Subplots nebeneinander
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# ---- Körpergröße ----
groesse = df["Groesse"]
mu_g, sigma_g = groesse.mean(), groesse.std()
# Histogramm
ax1.hist(groesse, bins=20, color="blue", alpha=0.7, rwidth=0.9, density=True)
# Normalverteilung überlagern
x_g = np.linspace(groesse.min(), groesse.max(), 100)
ax1.plot(x_g, norm.pdf(x_g, mu_g, sigma_g), color="darkred", linestyle="-", lw=3, label=f"N(μ={mu_g:.1f}, σ={sigma_g:.1f})")
ax1.set_title("Körpergröße")
ax1.set_xlabel("Größe (cm)")
ax1.set_ylabel("Dichte")
ax1.legend()

# ---- Gewicht ----
gewicht = df["Gewicht"]
mu_w, sigma_w = gewicht.mean(), gewicht.std()
# Histogramm
ax2.hist(gewicht, bins=20, color="green", alpha=0.7, rwidth=0.9, density=True)
# Normalverteilung überlagern
x_w = np.linspace(gewicht.min(), gewicht.max(), 100)
ax2.plot(x_w, norm.pdf(x_w, mu_w, sigma_w), color="darkred", linestyle="-", lw=3, label=f"N(μ={mu_w:.1f}, σ={sigma_w:.1f})")
ax2.set_title("Gewicht")
ax2.set_xlabel("Gewicht (kg)")
ax2.set_ylabel("Dichte")
ax2.legend()

plt.suptitle("Verteilung von Körpergröße und Gewicht")
plt.tight_layout()
```

```python
n = 5  # Anzahl Würfel pro Experiment
experimente = 1000
wuerfe = np.random.randint(1, 7, size=(experimente, n))
summen = wuerfe.sum(axis=1)
print(f"Der Mittelwert ist {summen.mean()} und die Standardabweichung ist {summen.std()}.")
```

```python
mu, sigma = summen.mean(), summen.std()
mu_babies, sigma_babies = 50, 2

# x-Werte definieren
x = np.linspace(mu-5*sigma, mu+5*sigma, 1000)

# Normalverteilungen berechnen
pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

# Plot
plt.plot(x, pdf, label='Dichtefunktion')
plt.xlabel("Würfelsumme (cm)")
plt.ylabel("Häufigkeit")
plt.title("Approximation der Würfelsummen durch Normalverteilung")
plt.legend(fontsize=10)
```

```python
# Plot
plt.plot(x, pdf, label='Dichtefunktion')

w = 5*n

# Fläche unterhalb der Funktion ab x = 10
x_fill = x[x >= w]
pdf_fill = pdf[x >= w]
plt.fill_between(x_fill, pdf_fill, color='red', alpha=0.4, label=f'P(X ≥ {w})')

# Pfeil + Text
plt.annotate(
    f'Wahrscheinlichkeit, dass eine\nWürfelsumme ≥ {w} auftritt',
    xy=(w+1, pdf[np.abs(x).argmin()]),
    xytext=(mu + sigma-.3, max(pdf) * 0.7),
    arrowprops=dict(facecolor='black', arrowstyle='->'),
    fontsize=10
)

plt.xlabel("Würfelsumme (cm)")
plt.ylabel("Dichte")
plt.title(f"(Theoretische) Verteilung der Würfelsummen (n={n})")
plt.legend(fontsize=10)
```

```python
dx = x[1] - x[0]
bereich = x >= w
fläche = np.sum(pdf[bereich]) * dx
print(f"P(X ≥ {w}) ≈ {fläche:.4f}")
```

```python
dx = x[1] - x[0]
cdf = np.cumsum(pdf) * dx  # Numerische Integration

plt.plot(x, cdf, label="Verteilungsfunktion (numerisch)")
plt.xlabel("Würfelsumme")
plt.ylabel("F(x)")
plt.title("Kumulative Verteilungsfunktion")
plt.grid(True)
```

```python
x_wert = 10
cdf_wert = np.interp(x_wert, x, cdf)
print(f"P(X ≥ {x_wert}) ≈ {1-cdf_wert:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Zufallsdaten mit linearem Zusammenhang erzeugen
np.random.seed(42)
x = np.random.normal(loc=50, scale=10, size=100)
y = 2 * x + np.random.normal(loc=0, scale=10, size=100)

# Korrelationskoeffizient berechnen
r, _ = pearsonr(x, y)

# Streudiagramm
plt.figure(figsize=(6, 4))
plt.scatter(x, y, alpha=0.7, edgecolor='k')
plt.title(f"Streudiagramm mit Pearson-Korrelation r = {r:.2f}")
plt.xlabel("Merkmal X")
plt.ylabel("Merkmal Y")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Daten erzeugen
np.random.seed(42)
x = np.random.normal(loc=50, scale=10, size=100)
y = 2 * x + np.random.normal(loc=0, scale=10, size=100)

# Mittelwerte berechnen
x_mean = np.mean(x)
y_mean = np.mean(y)

# Schritt 1: Originale Punktwolke
fig1, ax1 = plt.subplots(figsize=(6, 4))
ax1.scatter(x, y, alpha=0.6)
ax1.axvline(x_mean, color='gray', linestyle='--')
ax1.axhline(y_mean, color='gray', linestyle='--')
ax1.set_title("Originale Punktwolke\nmit Mittelwerten")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
ax1.grid(True)

# Schritt 2: Verschiebung in den Ursprung
x_centered = x - x_mean
y_centered = y - y_mean

fig2, ax2 = plt.subplots(figsize=(6, 4))
ax2.scatter(x_centered, y_centered, alpha=0.6)
ax2.axhline(0, color='gray', linestyle='--')
ax2.axvline(0, color='gray', linestyle='--')
ax2.set_title("Zentrierte Punktwolke\n(um Ursprung verschoben)")
ax2.set_xlabel("x - mean(x)")
ax2.set_ylabel("y - mean(y)")
ax2.grid(True)

# Schritt 3: Punkte nach Produkt der Abweichungen einfärben
weights = x_centered * y_centered
colors = np.where(weights >= 0, 'blue', 'red')

fig3, ax3 = plt.subplots(figsize=(6, 4))
ax3.scatter(x_centered, y_centered, c=colors, alpha=0.7, edgecolor='k')
ax3.axhline(0, color='gray', linestyle='--')
ax3.axvline(0, color='gray', linestyle='--')
ax3.set_title("Gewichtete Punkte\nblau = positiv, rot = negativ")
ax3.set_xlabel("x - mean(x)")
ax3.set_ylabel("y - mean(y)")
ax3.grid(True)

# Schritt 4: Korrelationskoeffizient berechnen
numerator = np.sum(x_centered * y_centered)
denominator = np.sqrt(np.sum(x_centered**2)) * np.sqrt(np.sum(y_centered**2))
r = numerator / denominator

print(f"Korrelationskoeffizient r = {r:.2f}")
```
